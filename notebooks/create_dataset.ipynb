```python
def show(path, depth=0, max_depth=3):
    if depth > max_depth:
        return
    try:
        items = sorted(os.listdir(path))
    except Exception as e:
        print("  " * depth + f"[ERR] {path}: {e}")
        return
    print("  " * depth + f"{path}/  ({len(items)} items)")
    if depth == max_depth:
        return
    for it in items[:5]:
        full = os.path.join(path, it)
        if os.path.isdir(full):
            show(full, depth + 1, max_depth)
        else:
            print("  " * (depth + 1) + it)

print("=== /kaggle/input ===")
show("/kaggle/input", 0, 2)

print()
print("=== /kaggle/input/binance-spot-klines-usdt ===")
show("/kaggle/input/binance-spot-klines-usdt", 0, 3)

print()
print("=== symlink check ===")
dst = "/kaggle/working/FinancAIBot/upload/data/spot"
print("islink:", os.path.islink(dst))
print("exists:", os.path.exists(dst))
if os.path.islink(dst):
    print("points to:", os.readlink(dst))
    print("resolved:", os.path.realpath(dst))
    print("listdir:", os.listdir(dst)[:10] if os.path.isdir(dst) else "N/A")
```

```python
import numpy as np

DATA_DIR = "/kaggle/working/FinancAIBot/upload/level_data/breakout_dataset"
file_paths = sorted(glob.glob(os.path.join(DATA_DIR, "*_breakout_levels.parquet")))
full_df = pd.concat([pd.read_parquet(p) for p in file_paths], ignore_index=True)

UNHASHABLE = ["touch_times"]
cols_for_dup = [c for c in full_df.columns if c not in UNHASHABLE]

print("=" * 70)
print("BASIC SHAPE")
print("=" * 70)
print(f"rows:            {len(full_df):,}")
print(f"tickers:         {full_df['ticker'].nunique()}")
print(f"unique levels:   {full_df['level_id'].nunique()}")
print(f"rows / level:    {len(full_df) / full_df['level_id'].nunique():.2f}")
print(f"time span:       {full_df['prediction_time'].min()} -> {full_df['prediction_time'].max()}")
print()

print("=" * 70)
print("1. NaN / INF / CONSTANT COLUMNS")
print("=" * 70)
num = full_df.select_dtypes(include=[np.number])
nan_cnt = num.isna().sum()
inf_cnt = np.isinf(num).sum()
const = [c for c in num.columns if num[c].nunique(dropna=False) <= 1]
print("columns with NaN:  ", nan_cnt[nan_cnt > 0].to_dict() or "none")
print("columns with Inf:  ", inf_cnt[inf_cnt > 0].to_dict() or "none")
print("constant columns:  ", const or "none")
print()

print("=" * 70)
print("2. DUPLICATES")
print("=" * 70)
dup_full = full_df.duplicated(subset=cols_for_dup).sum()
dup_keys = full_df.duplicated(subset=["ticker", "level_id", "prediction_time"]).sum()
print(f"full-row duplicates (excl. {UNHASHABLE}):  {dup_full}")
print(f"duplicates on (ticker, level_id, prediction_time): {dup_keys}")
print()

print("=" * 70)
print("3. LEVEL ID UNIQUENESS ACROSS TICKERS")
print("=" * 70)
lvl = full_df.groupby("level_id")["ticker"].nunique()
print("levels belonging to >1 ticker:", (lvl > 1).sum())
print()

print("=" * 70)
print("4. LABEL BALANCE (per horizon)")
print("=" * 70)
for c in ["breakout_1h", "breakout_4h", "breakout_24h", "is_confirmation_touch"]:
    vc = full_df[c].value_counts(dropna=False).sort_index()
    pos = int(vc.get(1, vc.get(True, 0)))
    neg = int(vc.get(0, vc.get(False, 0)))
    print(f"{c:22s}  neg={neg:6d}  pos={pos:6d}  pos_rate={pos/len(full_df):.4f}")
print()

print("=" * 70)
print("5. LABEL CONSISTENCY (monotonicity)")
print("=" * 70)
bad_24_4 = ((full_df["breakout_24h"] == 1) & (full_df["breakout_4h"] == 0)).sum()
bad_4_1  = ((full_df["breakout_4h"]  == 1) & (full_df["breakout_1h"] == 0)).sum()
bad_24_1 = ((full_df["breakout_24h"] == 1) & (full_df["breakout_1h"] == 0)).sum()
print(f"24h=1 & 4h=0:  {bad_24_4}")
print(f"4h=1  & 1h=0:  {bad_4_1}")
print(f"24h=1 & 1h=0:  {bad_24_1}")
print("(если > 0 — либо определения не вложенные, либо баг; проверьте)")
print()

print("=" * 70)
print("6. TIME ORDER SANITY")
print("=" * 70)
full_df["_pt"] = pd.to_datetime(full_df["prediction_time"], utc=True, errors="coerce")
full_df["_ft"] = pd.to_datetime(full_df["first_touch_time"], utc=True, errors="coerce")
full_df["_lt"] = pd.to_datetime(full_df["last_touch_time"], utc=True, errors="coerce")
print("bad prediction_time parse:", full_df["_pt"].isna().sum())
print("first_touch > prediction_time:", (full_df["_ft"] > full_df["_pt"]).sum())
print("last_touch  > prediction_time:", (full_df["_lt"] > full_df["_pt"]).sum())
print("first_touch > last_touch:     ", (full_df["_ft"] > full_df["_lt"]).sum())
print()

print("=" * 70)
print("7. FEATURE RANGES (sanity)")
print("=" * 70)
checks = [
    ("zone_low <= center <= zone_high",
        ((full_df["zone_low"] <= full_df["center"]) &
         (full_df["center"]  <= full_df["zone_high"])).mean()),
    ("zone_width_pct >= 0",       (full_df["zone_width_pct"] >= 0).mean()),
    ("zone_width_atr >= 0",       (full_df["zone_width_atr"] >= 0).mean()),
    ("touch_count >= 1",          (full_df["touch_count"] >= 1).mean()),
    ("recent_touch_count <= touch_count",
        (full_df["recent_touch_count"] <= full_df["touch_count"]).mean()),
    ("price_position_in_zone in [0,1]",
        full_df["price_position_in_zone"].between(0, 1).mean()),
    ("atr_pct >= 0",              (full_df["atr_pct"] >= 0).mean()),
    ("btc_return_* finite",       np.isfinite(full_df["btc_return_24h"]).mean()),
]
for name, frac in checks:
    flag = "OK" if frac > 0.99 else "CHECK"
    print(f"[{flag}] {name:45s} {frac:.4f}")
print()

print("=" * 70)
print("8. PER-TICKER COVERAGE")
print("=" * 70)
per = full_df.groupby("ticker").agg(
    rows=("level_id", "size"),
    levels=("level_id", "nunique"),
    pos_24h=("breakout_24h", "sum"),
    pos_rate_24h=("breakout_24h", "mean"),
).sort_values("rows")
print(per.to_string())
print()
print("tickers with < 50 rows:", (per["rows"] < 50).sum())
print("tickers with 0 positives (24h):", (per["pos_24h"] == 0).sum())
print("tickers with 0 negatives (24h):", (per["pos_24h"] == per["rows"]).sum())
print()

full_df.drop(columns=["_pt", "_ft", "_lt"], inplace=True)
```
